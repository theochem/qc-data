```python
"""Example of retrieving files from a Zenodo dataset."""

from pathlib import Path

from qc_data.providers.zenodo import ZenodoProvider
from qc_data.registry import DatasetRegistry

# Load the dataset registry.
registry_path = Path("zenodo_dataset.toml")
registry = DatasetRegistry.from_toml(registry_path.read_text(encoding="utf-8"))

dataset = registry.datasets["footstep_obesity"]
cache_dir = Path("qc_data_cache") / registry.cache["namespace"]

provider = ZenodoProvider()

# List files available in the dataset.
files = provider.list(dataset, recursive=True)
print(files)
```
